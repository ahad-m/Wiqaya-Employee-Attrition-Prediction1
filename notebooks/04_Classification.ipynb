## Label encoding

```python
X_train = pd.read_csv("../data/processed/X_train_label.csv")
X_test  = pd.read_csv("../data/processed/X_test_label.csv")
y_train = pd.read_csv("../data/processed/y_train.csv")
y_test  = pd.read_csv("../data/processed/y_test.csv")

print("X_train:", X_train.shape, "X_test:", X_test.shape)
print("y_train:", y_train.shape, "y_test:", y_test.shape)
print("Classes:", np.unique(y_train, return_counts=True))
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scoring = {
    "accuracy": "accuracy",
    "f1": "f1",
    "roc_auc": "roc_auc"
}
```

```python
models = {
    "LogisticRegression": LogisticRegression(
        max_iter=5000,
        class_weight="balanced",
        solver="liblinear"
    ),
    
    "XGBoost": XGBClassifier(
        n_estimators=800,
        learning_rate=0.01,
        max_depth=3,
        subsample=1.0,
        colsample_bytree=0.7,
        random_state=42,
        n_jobs=-1
    )
}
```

```python
for name, model in models.items():
    
    print("\n" + "="*60)
    print("Final Test —", name)

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))

    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)[:, 1]
        print("Test ROC-AUC:", roc_auc_score(y_test, y_proba))
```

## Target Encoding with feautre engineering

```python
import numpy as np
import pandas as pd

# Load your FE1 files
X_train = pd.read_csv("../data/processed/X_train_fe1.csv")
X_test  = pd.read_csv("../data/processed/X_test_fe1.csv")

y_train = pd.read_csv("../data/processed/y_train_fe1.csv").squeeze()
y_test  = pd.read_csv("../data/processed/y_test_fe1.csv").squeeze()

cat_cols = ["Sector", "Department", "JobTitle"]  
cat_cols = [c for c in cat_cols if c in X_train.columns]  

def target_encode_train_test(X_tr, y_tr, X_te, cols, smoothing=10):
    X_tr = X_tr.copy()
    X_te = X_te.copy()

    global_mean = y_tr.mean()

    for col in cols:
        tr_col = X_tr[col].astype(str)
        te_col = X_te[col].astype(str)

        stats = y_tr.groupby(tr_col).agg(["mean", "count"])
        means = stats["mean"]
        counts = stats["count"]

        # smoothing: (count*mean + smoothing*global) / (count+smoothing)
        smooth = (counts * means + smoothing * global_mean) / (counts + smoothing)

        X_tr[col] = tr_col.map(smooth).fillna(global_mean)
        X_te[col] = te_col.map(smooth).fillna(global_mean)

    return X_tr, X_te

X_train_te, X_test_te = target_encode_train_test(X_train, y_train, X_test, cat_cols, smoothing=10)

print("After TE shapes:", X_train_te.shape, X_test_te.shape)
print("Encoded cols:", cat_cols)

# Save
X_train_te.to_csv("../data/processed/X_train_fe1_target.csv", index=False)
X_test_te.to_csv("../data/processed/X_test_fe1_target.csv", index=False)

print("✅ Saved: X_train_fe1_target.csv, X_test_fe1_target.csv")
```

## Target encoding modeling

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, classification_report, confusion_matrix

from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
```

```python
X_train = pd.read_csv("../data/processed/X_train_fe1_target.csv")
X_test  = pd.read_csv("../data/processed/X_test_fe1_target.csv")
print("X_train:", X_train.shape, "X_test:", X_test.shape)
print("y_train:", y_train.shape, "y_test:", y_test.shape)
print("Classes:", np.unique(y_train, return_counts=True))
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scoring = {
    "accuracy": "accuracy",
    "f1": "f1",
    "roc_auc": "roc_auc"
}
```

```python
models = {
    "LogisticRegression": LogisticRegression(
        max_iter=5000,
        class_weight="balanced",
        solver="liblinear"
    ),
    
    "XGBoost": XGBClassifier(
        n_estimators=800,
        learning_rate=0.01,
        max_depth=3,
        subsample=1.0,
        colsample_bytree=0.7,
        random_state=42,
        n_jobs=-1
    )
}
```

```python
for name, model in models.items():
    
    print("\n" + "="*60)
    print("Final Test —", name)

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))

    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)[:, 1]
        print("Test ROC-AUC:", roc_auc_score(y_test, y_proba))
```

```python
import os
import joblib
from datetime import datetime
from sklearn.metrics import roc_auc_score, classification_report, accuracy_score, f1_score

os.makedirs("models", exist_ok=True)

best_name = None
best_model = None
best_score = -1

for name, model in models.items():
    print("\n" + "="*60)
    print("Final Test —", name)

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))

    # اختر المقياس اللي تبي تقارن عليه
    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)[:, 1]
        score = roc_auc_score(y_test, y_proba)
        print("Test ROC-AUC:", score)
    else:
        score = f1_score(y_test, y_pred)   # بديل لو ما فيه predict_proba
        print("Test F1:", score)

    if score > best_score:
        best_score = score
        best_name = name
        best_model = model

# حفظ أفضل مودل
ts = datetime.now().strftime("%Y%m%d_%H%M%S")
save_path = f"models/best_model_{best_name}_{best_score:.4f}_{ts}.joblib"
joblib.dump(best_model, save_path)

print("\n" + "="*60)
print(f"✅ Best model: {best_name} | score={best_score:.4f}")
print(f"💾 Saved to: {save_path}")
```
